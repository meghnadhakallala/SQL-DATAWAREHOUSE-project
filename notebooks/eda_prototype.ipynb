# EDA -- IoT sensor telemetry

Exploratory look at the gold-layer engine sensor data before/after modeling.
Run the pipeline scripts in `src/` first so `iot_warehouse.db` exists.

```python
import pandas as pd
import sqlalchemy
import matplotlib.pyplot as plt

engine = sqlalchemy.create_engine('sqlite:///../iot_warehouse.db')
df = pd.read_sql('SELECT * FROM gold_engine_metrics', con=engine)
df.head()
```

```python
# Which engines are closest to failure right now?
df.groupby('engine_id')['failure_within_30_cycles'].mean().sort_values(ascending=False)
```

```python
sample = df[df['engine_id'] == df['engine_id'].iloc[0]]
plt.figure(figsize=(8, 4))
plt.plot(sample['cycle'], sample['vibration'], label='raw vibration')
plt.plot(sample['cycle'], sample['rolling_avg_vibration'], label='rolling avg')
plt.legend()
plt.title('Vibration trend')
plt.xlabel('Operating cycle')
plt.show()
```
